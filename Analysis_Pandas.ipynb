```python
#Install datasets Library
# !pip install datasets

#Import Library
import pandas as pd
from datasets import load_dataset

# Loading DATA
datasets = load_dataset('lukebarousse/data_jobs')
df = datasets['train'].to_pandas()

#Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df.groupby('job_title_short')['salary_year_avg'].median() # using group by method
```

```python
df.groupby(['job_title_short','job_country'])['salary_year_avg'].median()
```

Here's an example demonstrating how to apply multiple aggregation functions at once. We'll find the `min`, `max`, and `median` `salary_year_avg` for each `job_title_short`.

```python
aggregated_salaries = df.groupby('job_title_short')['salary_year_avg'].agg(['min', 'max', 'median'])
display(aggregated_salaries)
```

You can also apply different functions to different columns, or even custom functions. For example, let's get the median salary and the count of job postings for each `job_title_short`.

```python
custom_agg = df.groupby('job_title_short').agg(
    median_salary=('salary_year_avg', 'median'),
    job_post_count=('job_title_short', 'count')
)
display(custom_agg)
```
