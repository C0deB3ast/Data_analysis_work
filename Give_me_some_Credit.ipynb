```python
df = df.rename(columns={'Unnamed: 0' : 'Customer_Id'})
```

```python
median_income = df['MonthlyIncome'].median()
df['MonthlyIncome'] = df['MonthlyIncome'].fillna(median_income)
```

```python
mode_dependents = df['NumberOfDependents'].mode()[0]
df['NumberOfDependents'] = df['NumberOfDependents'].fillna(mode_dependents)
```

```python
df.isna().sum()
```

```python
median_age = df['age'].median()
df.loc[df['age'] == 0, 'age'] = median_age
```

```python
df[df['NumberOfTimes90DaysLate'] >= 96]
```

```python
df[(df['NumberOfTime30-59DaysPastDueNotWorse'] >= 96) &
   (df['NumberOfTimes90DaysLate'] >= 96) &
   (df['NumberOfTime60-89DaysPastDueNotWorse'] >= 96)].shape
```

```python
df = df[~((df['NumberOfTime30-59DaysPastDueNotWorse'] >= 96) &
   (df['NumberOfTimes90DaysLate'] >= 96) &
   (df['NumberOfTime60-89DaysPastDueNotWorse'] >= 96))]
```

```python
df.shape
```

```python
df[df['RevolvingUtilizationOfUnsecuredLines']>1].shape
```

```python
df['RevolvingUtilizationOfUnsecuredLines'].sort_values(ascending=False).head(20)
```

```python
cap_val= df['RevolvingUtilizationOfUnsecuredLines'].quantile(0.99)
df['RevolvingUtilizationOfUnsecuredLines'] = df['RevolvingUtilizationOfUnsecuredLines'].clip(upper=cap_val)
```

```python
df['RevolvingUtilizationOfUnsecuredLines'].describe()
```

```python
debt_cap_val = df['DebtRatio'].quantile(0.99)
df['DebtRatio'] = df['DebtRatio'].clip(upper=debt_cap_val)
```

```python
df['DebtRatio'].describe()
```
