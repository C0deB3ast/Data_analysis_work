### Problem 1: Distribution of Yearly Salaries
Create a histogram to visualize the distribution of `salary_year_avg`. Make sure to drop the `NaN` values first so the plot renders correctly.

```python
import matplotlib.pyplot as plt

# Filter for rows where salary is not null
salary_data = df['salary_year_avg'].dropna()

plt.figure(figsize=(10, 6))
plt.hist(salary_data, bins=50, color='skyblue', edgecolor='black')
plt.title('Distribution of Average Yearly Salaries')
plt.xlabel('Salary (USD)')
plt.ylabel('Number of Jobs')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

### Problem 2: Top 10 Countries by Job Postings
Create a horizontal bar chart showing the top 10 countries (`search_location`) with the most job postings.

```python
top_locations = df['search_location'].value_counts().head(10).sort_values(ascending=True)

plt.figure(figsize=(10, 8))
plt.barh(top_locations.index, top_locations.values, color='salmon')
plt.title('Top 10 Locations for Data Jobs')
plt.xlabel('Count of Job Postings')
plt.ylabel('Location')
plt.tight_layout()
plt.show()
```

**Median Salary by job title**

```python
job_med_sal= df.groupby('job_title_short')['salary_year_avg'].median().sort_values()
job_med_sal
```
